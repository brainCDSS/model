# H-N00 · I-N00 · A-B00 nnU-Net baseline — held-out test export

Validation에서 선택된 best checkpoint를 바꾸지 않고 각 held-out test cohort에 full-volume inference한다.

- H-N00 CT: 40 cases
- I-N00 MRI: 74 cases
- A-B00 MRA: 46 cases

Output: `runs/final_test_exports/<MODEL_ID>/predictions/<case_id>.nii.gz`


```python
from pathlib import Path
import subprocess
REPO = Path("/home/stu03/workspace/model")
PYTHON = REPO / ".venv/bin/python"
EXPORTER = REPO / "scripts/evaluation/export_nnunet_baseline_tests.py"
assert PYTHON.is_file() and EXPORTER.is_file()
MODEL_IDS = ["H-N00", "I-N00", "A-B00"]
print("Will export:", MODEL_IDS)
```

## GPU / checkpoint smoke test (one case per model)

Run once first. It writes one valid original-space NIfTI mask for each model and confirms its custom checkpoint can be loaded.


```python
subprocess.run([str(PYTHON), str(EXPORTER), "--model-id", *MODEL_IDS, "--limit", "1"], cwd=REPO, check=True)
```

## Full held-out test export

After smoke success, run once. Completed masks are skipped automatically, so interrupted work resumes safely.


```python
subprocess.run([str(PYTHON), str(EXPORTER), "--model-id", *MODEL_IDS], cwd=REPO, check=True)
```

```python
for model_id in MODEL_IDS:
    prediction_dir = REPO / "runs/final_test_exports" / model_id / "predictions"
    print(model_id, len(list(prediction_dir.glob("*.nii.gz"))), prediction_dir)
```
